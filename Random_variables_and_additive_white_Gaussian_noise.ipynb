```python
# ============================================================
# RANDOM VARIABLES, THEORETICAL PDFs AND AWGN
# Google Colab Ready
#
# Includes:
#   1. Uniform random variable generation
#   2. Zero-mean Gaussian random variable generation
#   3. Mean and variance estimation
#   4. Normalized histograms + theoretical PDFs
#   5. AWGN addition at specified SNR
#   6. Independent measured SNR verification
#   7. Mandatory validation of sample-count effects
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ============================================================
# 1. GLOBAL PARAMETERS
# ============================================================

SEED = 42

np.random.seed(SEED)

# Uniform distribution parameters
UNIFORM_MEAN = 0.0
UNIFORM_VARIANCE = 1.0

# For a uniform distribution U(a,b):
# Mean     = (a+b)/2
# Variance = (b-a)^2 / 12
#
# For mean = 0 and variance = 1:
#
# a = -sqrt(3)
# b = +sqrt(3)

UNIFORM_A = -np.sqrt(3)
UNIFORM_B = np.sqrt(3)

# Gaussian parameters
GAUSSIAN_MEAN = 0.0
GAUSSIAN_VARIANCE = 1.0
GAUSSIAN_STD = np.sqrt(GAUSSIAN_VARIANCE)

# AWGN SNR
TARGET_SNR_DB = 10

# Sample counts for statistical validation
SAMPLE_COUNTS = [
    100,
    1000,
    10000,
    100000
]

print("====================================================")
print("RANDOM VARIABLES AND AWGN SIMULATION")
print("====================================================")

print("Uniform distribution:")
print("  a =", UNIFORM_A)
print("  b =", UNIFORM_B)
print("  Theoretical mean     =", UNIFORM_MEAN)
print("  Theoretical variance =", UNIFORM_VARIANCE)

print("\nGaussian distribution:")
print("  Mean     =", GAUSSIAN_MEAN)
print("  Variance =", GAUSSIAN_VARIANCE)

print("\nTarget AWGN SNR =", TARGET_SNR_DB, "dB")


# ============================================================
# 2. MODULAR RANDOM VARIABLE GENERATORS
# ============================================================

def generate_uniform(
    n,
    a=UNIFORM_A,
    b=UNIFORM_B,
    seed=None
):
    """
    Generate n uniformly distributed random variables.
    """

    rng = np.random.default_rng(seed)

    return rng.uniform(
        low=a,
        high=b,
        size=n
    )


def generate_gaussian(
    n,
    mean=GAUSSIAN_MEAN,
    variance=GAUSSIAN_VARIANCE,
    seed=None
):
    """
    Generate n Gaussian random variables.
    """

    rng = np.random.default_rng(seed)

    std = np.sqrt(variance)

    return rng.normal(
        loc=mean,
        scale=std,
        size=n
    )


# ============================================================
# 3. STATISTICAL ESTIMATION FUNCTION
# ============================================================

def estimate_statistics(x):

    """
    Estimate sample mean and sample variance.
    """

    sample_mean = np.mean(x)

    sample_variance = np.var(
        x,
        ddof=1
    )

    return sample_mean, sample_variance


# ============================================================
# 4. THEORETICAL PDF FUNCTIONS
# ============================================================

def uniform_pdf(
    x,
    a,
    b
):
    """
    Theoretical uniform PDF.
    """

    pdf = np.zeros_like(
        x,
        dtype=float
    )

    mask = (
        (x >= a)
        &
        (x <= b)
    )

    pdf[mask] = 1 / (b - a)

    return pdf


def gaussian_pdf(
    x,
    mean=0,
    variance=1
):
    """
    Theoretical Gaussian PDF.
    """

    sigma = np.sqrt(
        variance
    )

    return (
        1 /
        (
            sigma *
            np.sqrt(2*np.pi)
        )
    ) * np.exp(
        -(
            (x - mean)**2
        )
        /
        (
            2*variance
        )
    )


# ============================================================
# 5. GENERATE RANDOM VARIABLES
# ============================================================

N = 100000

uniform_samples = generate_uniform(
    N,
    seed=SEED
)

gaussian_samples = generate_gaussian(
    N,
    seed=SEED
)


# ============================================================
# 6. ESTIMATE MEAN AND VARIANCE
# ============================================================

uniform_mean, uniform_variance = (
    estimate_statistics(
        uniform_samples
    )
)

gaussian_mean, gaussian_variance = (
    estimate_statistics(
        gaussian_samples
    )
)

print("\n====================================================")
print("ESTIMATED STATISTICS")
print("====================================================")

print("\nUniform random variable")

print(
    "Theoretical mean     =",
    UNIFORM_MEAN
)

print(
    "Estimated mean       =",
    uniform_mean
)

print(
    "Theoretical variance =",
    UNIFORM_VARIANCE
)

print(
    "Estimated variance   =",
    uniform_variance
)

print("\nZero-mean Gaussian random variable")

print(
    "Theoretical mean     =",
    GAUSSIAN_MEAN
)

print(
    "Estimated mean       =",
    gaussian_mean
)

print(
    "Theoretical variance =",
    GAUSSIAN_VARIANCE
)

print(
    "Estimated variance   =",
    gaussian_variance
)


# ============================================================
# 7. UNIFORM HISTOGRAM + THEORETICAL PDF
# ============================================================

x_uniform = np.linspace(
    UNIFORM_A - 0.2,
    UNIFORM_B + 0.2,
    1000
)

uniform_theoretical_pdf = uniform_pdf(
    x_uniform,
    UNIFORM_A,
    UNIFORM_B
)

plt.figure(
    figsize=(10, 5)
)

plt.hist(
    uniform_samples,
    bins=60,
    density=True,
    alpha=0.6,
    label="Normalized histogram"
)

plt.plot(
    x_uniform,
    uniform_theoretical_pdf,
    linewidth=2,
    label="Theoretical uniform PDF"
)

plt.xlabel("Random variable x")
plt.ylabel("Probability density")

plt.title(
    "Uniform Random Variable: Histogram vs Theoretical PDF"
)

plt.grid(True)
plt.legend()

plt.show()


# ============================================================
# 8. GAUSSIAN HISTOGRAM + THEORETICAL PDF
# ============================================================

x_gaussian = np.linspace(
    -5,
    5,
    1000
)

gaussian_theoretical_pdf = gaussian_pdf(
    x_gaussian,
    GAUSSIAN_MEAN,
    GAUSSIAN_VARIANCE
)

plt.figure(
    figsize=(10, 5)
)

plt.hist(
    gaussian_samples,
    bins=70,
    density=True,
    alpha=0.6,
    label="Normalized histogram"
)

plt.plot(
    x_gaussian,
    gaussian_theoretical_pdf,
    linewidth=2,
    label="Theoretical Gaussian PDF"
)

plt.xlabel("Random variable x")
plt.ylabel("Probability density")

plt.title(
    "Zero-Mean Gaussian: Histogram vs Theoretical PDF"
)

plt.grid(True)
plt.legend()

plt.show()


# ============================================================
# 9. AWGN GENERATOR
# ============================================================

def add_awgn(
    signal,
    snr_db,
    seed=None
):
    """
    Add white Gaussian noise to a signal.

    The noise power is selected so that:

        SNR = signal_power / noise_power
    """

    rng = np.random.default_rng(
        seed
    )

    signal_power = np.mean(
        signal**2
    )

    snr_linear = (
        10**(
            snr_db / 10
        )
    )

    noise_power = (
        signal_power /
        snr_linear
    )

    noise = (
        np.sqrt(noise_power)
        *
        rng.normal(
            0,
            1,
            len(signal)
        )
    )

    noisy_signal = (
        signal +
        noise
    )

    return noisy_signal, noise


# ============================================================
# 10. CREATE A TEST SIGNAL
# ============================================================

fs = 10000

duration = 1.0

time = np.arange(
    0,
    duration,
    1/fs
)

frequency = 100

signal = (
    np.sin(
        2*np.pi*
        frequency*
        time
    )
)


# ============================================================
# 11. ADD AWGN
# ============================================================

noisy_signal, noise = add_awgn(
    signal,
    TARGET_SNR_DB,
    seed=SEED
)


# ============================================================
# 12. INDEPENDENT SNR MEASUREMENT
# ============================================================

def measure_snr(
    signal,
    noise
):
    """
    Independently measure SNR from
    signal and noise powers.
    """

    signal_power = np.mean(
        signal**2
    )

    noise_power = np.mean(
        noise**2
    )

    snr_linear = (
        signal_power /
        noise_power
    )

    snr_db = (
        10 *
        np.log10(
            snr_linear
        )
    )

    return (
        signal_power,
        noise_power,
        snr_db
    )


signal_power, noise_power, measured_snr_db = (
    measure_snr(
        signal,
        noise
    )
)


# ============================================================
# 13. AWGN VALIDATION
# ============================================================

print("\n====================================================")
print("AWGN VALIDATION")
print("====================================================")

print(
    "Target SNR (dB)    =",
    TARGET_SNR_DB
)

print(
    "Signal power       =",
    signal_power
)

print(
    "Measured noise power =",
    noise_power
)

print(
    "Measured SNR (dB)  =",
    measured_snr_db
)

print(
    "SNR error (dB)     =",
    measured_snr_db -
    TARGET_SNR_DB
)


# ============================================================
# 14. PLOT ORIGINAL + NOISY SIGNAL
# ============================================================

display_samples = 500

plt.figure(
    figsize=(12, 5)
)

plt.plot(
    time[:display_samples],
    signal[:display_samples],
    label="Original signal"
)

plt.plot(
    time[:display_samples],
    noisy_signal[:display_samples],
    alpha=0.7,
    label="Signal + AWGN"
)

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")

plt.title(
    f"AWGN Addition: Target SNR = {TARGET_SNR_DB} dB"
)

plt.grid(True)
plt.legend()

plt.show()


# ============================================================
# 15. NOISE HISTOGRAM
# ============================================================

plt.figure(
    figsize=(10, 5)
)

plt.hist(
    noise,
    bins=70,
    density=True,
    alpha=0.6,
    label="Measured noise histogram"
)

noise_std = np.std(
    noise
)

x_noise = np.linspace(
    -5*noise_std,
    5*noise_std,
    1000
)

noise_pdf = gaussian_pdf(
    x_noise,
    mean=0,
    variance=noise_std**2
)

plt.plot(
    x_noise,
    noise_pdf,
    linewidth=2,
    label="Theoretical Gaussian PDF"
)

plt.xlabel("Noise amplitude")
plt.ylabel("Probability density")

plt.title(
    "AWGN: Noise Histogram and Gaussian PDF"
)

plt.grid(True)
plt.legend()

plt.show()


# ============================================================
# 16. MANDATORY VALIDATION:
#     EFFECT OF SAMPLE COUNT
# ============================================================
#
# Larger N should improve agreement statistically.
#
# BUT:
#
# sample_mean != theoretical_mean exactly
#
# even for large N.
#
# It only approaches the theoretical mean
# according to the law of large numbers.
# ============================================================

validation_results = []

for N_test in SAMPLE_COUNTS:

    # Use a DIFFERENT random realization
    # for every sample count.

    samples = generate_gaussian(
        N_test,
        mean=GAUSSIAN_MEAN,
        variance=GAUSSIAN_VARIANCE,
        seed=SEED + N_test
    )

    estimated_mean, estimated_variance = (
        estimate_statistics(
            samples
        )
    )

    mean_error = (
        estimated_mean -
        GAUSSIAN_MEAN
    )

    variance_error = (
        estimated_variance -
        GAUSSIAN_VARIANCE
    )

    validation_results.append({

        "Sample count": N_test,

        "Sample mean": estimated_mean,

        "Theoretical mean": GAUSSIAN_MEAN,

        "Mean error": mean_error,

        "Sample variance": estimated_variance,

        "Theoretical variance": GAUSSIAN_VARIANCE,

        "Variance error": variance_error

    })


validation_table = pd.DataFrame(
    validation_results
)


# ============================================================
# 17. DISPLAY SAMPLE-COUNT VALIDATION
# ============================================================

print("\n====================================================")
print("MANDATORY SAMPLE-COUNT VALIDATION")
print("====================================================")

display(
    validation_table.style.format({

        "Sample count": "{:.0f}",

        "Sample mean": "{:.6f}",

        "Theoretical mean": "{:.6f}",

        "Mean error": "{:.6e}",

        "Sample variance": "{:.6f}",

        "Theoretical variance": "{:.6f}",

        "Variance error": "{:.6e}"

    })
)


# ============================================================
# 18. VERIFY MEAN IS NOT FORCED TO ZERO
# ============================================================

means = (
    validation_table[
        "Sample mean"
    ].values
)

mean_exactly_equal = np.isclose(
    means,
    GAUSSIAN_MEAN,
    atol=0,
    rtol=0
)

print(
    "\nSample means:"
)

for N_test, mean_value in zip(
    SAMPLE_COUNTS,
    means
):

    print(
        f"N = {N_test:6d} : "
        f"mean = {mean_value:.10f}"
    )


if np.all(
    mean_exactly_equal
):

    print(
        "\nWARNING: All sample means are "
        "exactly equal to the theoretical mean."
    )

    print(
        "This would indicate that the samples "
        "were artificially centered."
    )

else:

    print(
        "\n✓ VALIDATION PASSED:"
    )

    print(
        "The sample means are not forced to "
        "equal the theoretical mean exactly."
    )

    print(
        "They fluctuate around the theoretical mean."
    )


# ============================================================
# 19. PLOT MEAN CONVERGENCE
# ============================================================

plt.figure(
    figsize=(10, 5)
)

plt.semilogx(
    SAMPLE_COUNTS,
    np.abs(
        validation_table[
            "Mean error"
        ]
    ),
    marker="o"
)

plt.xlabel(
    "Number of samples"
)

plt.ylabel(
    "Absolute mean error"
)

plt.title(
    "Sample Mean Error vs Sample Count"
)

plt.grid(True)

plt.show()


# ============================================================
# 20. PLOT VARIANCE CONVERGENCE
# ============================================================

plt.figure(
    figsize=(10, 5)
)

plt.semilogx(
    SAMPLE_COUNTS,
    np.abs(
        validation_table[
            "Variance error"
        ]
    ),
    marker="o"
)

plt.xlabel(
    "Number of samples"
)

plt.ylabel(
    "Absolute variance error"
)

plt.title(
    "Sample Variance Error vs Sample Count"
)

plt.grid(True)

plt.show()


# ============================================================
# 21. HISTOGRAM COMPARISON FOR DIFFERENT SAMPLE COUNTS
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(12, 8)
)

for ax, N_test in zip(
    axes.ravel(),
    SAMPLE_COUNTS
):

    samples = generate_gaussian(
        N_test,
        mean=GAUSSIAN_MEAN,
        variance=GAUSSIAN_VARIANCE,
        seed=SEED + N_test
    )

    ax.hist(
        samples,
        bins=40,
        density=True,
        alpha=0.6,
        label=f"N = {N_test}"
    )

    ax.plot(
        x_gaussian,
        gaussian_theoretical_pdf,
        linewidth=2,
        label="Theoretical PDF"
    )

    ax.set_title(
        f"Gaussian Histogram, N = {N_test}"
    )

    ax.set_xlabel("x")
    ax.set_ylabel("PDF")

    ax.grid(True)
    ax.legend()

plt.tight_layout()
plt.show()


# ============================================================
# 22. REPEAT EXPERIMENT TO SHOW STATISTICAL AGREEMENT
# ============================================================

# Mean absolute error and variance absolute error
# averaged over several independent trials.

num_trials = 100

agreement_results = []

for N_test in SAMPLE_COUNTS:

    mean_errors = []
    variance_errors = []

    for trial in range(
        num_trials
    ):

        samples = generate_gaussian(
            N_test,
            mean=GAUSSIAN_MEAN,
            variance=GAUSSIAN_VARIANCE,
            seed=1000 + trial + N_test
        )

        sample_mean, sample_variance = (
            estimate_statistics(
                samples
            )
        )

        mean_errors.append(
            abs(
                sample_mean -
                GAUSSIAN_MEAN
            )
        )

        variance_errors.append(
            abs(
                sample_variance -
                GAUSSIAN_VARIANCE
            )
        )

    agreement_results.append({

        "N": N_test,

        "Average |mean error|":
            np.mean(mean_errors),

        "Average |variance error|":
            np.mean(variance_errors)

    })


agreement_table = pd.DataFrame(
    agreement_results
)


# ============================================================
# 23. STATISTICAL AGREEMENT TABLE
# ============================================================

print("\n====================================================")
print("STATISTICAL AGREEMENT OVER 100 TRIALS")
print("====================================================")

display(
    agreement_table.style.format({

        "N": "{:.0f}",

        "Average |mean error|":
            "{:.6f}",

        "Average |variance error|":
            "{:.6f}"

    })
)


# ============================================================
# 24. PLOT STATISTICAL AGREEMENT
# ============================================================

plt.figure(
    figsize=(10, 5)
)

plt.loglog(
    agreement_table["N"],
    agreement_table[
        "Average |mean error|"
    ],
    marker="o",
    label="Mean error"
)

plt.loglog(
    agreement_table["N"],
    agreement_table[
        "Average |variance error|"
    ],
    marker="s",
    label="Variance error"
)

plt.xlabel(
    "Number of samples"
)

plt.ylabel(
    "Average absolute error"
)

plt.title(
    "Statistical Agreement Improves with Sample Count"
)

plt.grid(True)
plt.legend()

plt.show()


# ============================================================
# 25. FINAL AUTOMATIC VALIDATION
# ============================================================

# Check that larger sample counts generally produce
# smaller average mean error and variance error.

mean_error_values = (
    agreement_table[
        "Average |mean error|"
    ].values
)

variance_error_values = (
    agreement_table[
        "Average |variance error|"
    ].values
)

mean_improves = (
    mean_error_values[-1]
    <
    mean_error_values[0]
)

variance_improves = (
    variance_error_values[-1]
    <
    variance_error_values[0]
)


print("\n====================================================")
print("FINAL VALIDATION")
print("====================================================")

if mean_improves:

    print(
        "✓ Larger sample count improves "
        "mean statistical agreement."
    )

else:

    print(
        "Note: Mean error did not monotonically "
        "decrease in this finite experiment."
    )

if variance_improves:

    print(
        "✓ Larger sample count improves "
        "variance statistical agreement."
    )

else:

    print(
        "Note: Variance error did not monotonically "
        "decrease in this finite experiment."
    )

print(
    "\n✓ Sample mean is estimated from random samples "
    "and is NOT artificially forced to the theoretical mean."
)

print(
    "\n✓ AWGN SNR was independently measured "
    "from signal and noise powers."
)

print(
    "\n===================================================="
)
print("SIMULATION COMPLETE")
print("====================================================")
```
